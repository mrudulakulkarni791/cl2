```python
import nltk
import re
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, SnowballStemmer, WordNetLemmatizer
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
text = """The Students are studying Information Retrieval.
They are searching for useful information and retrieving documents.
Pre-processing removes unnecessary words, numbers and symbols. 123"""
print(text)
```

```python
text = text.lower()
print(text)
```

```python
text = re.sub(r'\d+', '', text)
print(text)
```

```python
text = re.sub(r'[^\w\s]', '', text)
print(text)
```

```python
words = word_tokenize(text)
print(words)
```

```python
stop_words = set(stopwords.words('english'))

filtered_words = [w for w in words if w not in stop_words]

print(filtered_words)
```

```python
porter = PorterStemmer()

porter_words = [porter.stem(w) for w in filtered_words]

print(porter_words)
```

```python
text_without_spaces = re.sub(r'\s', '', text)

print(text_without_spaces)
```

```python
snowball = SnowballStemmer('english')

snowball_words = [snowball.stem(w) for w in filtered_words]

print(snowball_words)
```

```python
lemmatizer = WordNetLemmatizer()

lemma_words = [lemmatizer.lemmatize(w) for w in filtered_words]

print(lemma_words)
```

```python
clean_text = ' '.join(filtered_words)

print("Final Preprocessed Text:")
print(clean_text)
```
