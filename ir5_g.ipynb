```python
import numpy as np
import networkx as nx
import matplotlib.pyplot as plt
```

```python
G = nx.DiGraph()

G.add_edges_from([
    ('B', 'A'),
    ('B', 'C'),
    ('C', 'A'),
    ('D', 'A'),
    ('D', 'B'),
    ('D', 'C')
])

print("Nodes:", G.nodes())
print("Edges:", G.edges())
```

```python
plt.figure(figsize=(7, 5))

nx.draw(
    G,
    with_labels=True,
    node_color='skyblue',
    node_size=2000,
    font_size=15,
    arrows=True
)

plt.title("Web Page Link Graph")
plt.show()
```

```python
pagerank = nx.pagerank(
    G,
    alpha=0.85,
    max_iter=100
)

print("PageRank Scores:")

for page, score in pagerank.items():
    print(page, ":", round(score, 4))
```

```python
ranked_pages = sorted(
    pagerank.items(),
    key=lambda x: x[1],
    reverse=True
)

print("Pages Ranked by Importance:")

for page, score in ranked_pages:
    print(page, ":", round(score, 4))
```

```python
pages = list(pagerank.keys())
scores = list(pagerank.values())

plt.figure(figsize=(8, 5))

plt.bar(pages, scores)

plt.title("PageRank Scores")
plt.xlabel("Web Pages")
plt.ylabel("PageRank Score")

plt.show()
```

```python
highest_page = max(pagerank, key=pagerank.get)

print("Highest Ranked Page:", highest_page)
print("Highest PageRank Score:", round(pagerank[highest_page], 4))
```
