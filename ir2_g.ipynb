```python
import re
```

```python
documents = {
    "Doc1": "The quick brown fox jumped over the lazy dog.",
    "Doc2": "The lazy dog slept in the sun.",
    "Doc3": "The quick dog runs in the garden."
}

for doc, text in documents.items():
    print(doc, ":", text)
```

```python
tokens = {}

for doc, text in documents.items():
    tokens[doc] = re.findall(r'\b\w+\b', text.lower())

print(tokens)
```

```python
stop_words = {"the", "is", "in", "over", "and", "a", "an"}

clean_tokens = {}

for doc, words in tokens.items():
    clean_tokens[doc] = [w for w in words if w not in stop_words]

print(clean_tokens)
```

```python
inverted_index = {}

for doc, words in clean_tokens.items():
    for word in words:
        if word not in inverted_index:
            inverted_index[word] = []

        if doc not in inverted_index[word]:
            inverted_index[word].append(doc)

print("Inverted Index:")

for word, docs in inverted_index.items():
    print(word, "->", docs)
```

```python
search_word = input("Enter word to search: ").lower()

if search_word in inverted_index:
    print("Document found:", inverted_index[search_word])
else:
    print("Word not found")
```
